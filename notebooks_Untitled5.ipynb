```python
# Exercise 1 : HANDLING ZERO DIVISION ERROR
def safe_divide():
    try:
        numerator = float(input("Enter numerator :"))
        denominator = float(input("Enter denominator:"))
        result = numerator/denominator
        print("Result:",result)
    except ZeroDivisionError:
        print("Error : Cannot divide by zero.")
        return None

        
safe_divide()
```

```python
# Exercise 2 : HANDLING VALUE ERROR
import math
def find_square_root(number1):
    try:
        if number1<0:
            raise ValueError("negative number provided")
            
        result = math.sqrt(number1)
        print("Square root:",result)
    except ValueError:
        print("Invalid input! Please enter a positive integer or a float.")
number1 = float(input("Enter a number:"))
find_square_root(number1)
```

```python
# Exercise 3 : HANDLING GENERIC EXCEPTIONS
def calculate(num):
    try:
        result =num/(num -5)
        print("Result:",result)
    except Exception:
        print("An error occured during calculation .")
num = float(input("Enter a number :"))
calculate(num)
```
