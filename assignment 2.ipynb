```python
# Create an empty list
shopping_list = []
```

```python
# Store items in a shopping list
shopping_list=["Watch","laptop","shoes","pen","clothes"]
print(shopping_list)
```

```python
# Add new item football to the shopping list
shopping_list.append("Football")
print(shopping_list)
```

```python
# Print last item from the shopping list
print("last item :", shopping_list[-1])
```

```python
# Print the entire shopping list
print("Shooping list :",shopping_list)
```

```python
# Print important items : laptop and shoes
print("Important items :",shopping_list[1],"and", shopping_list[2])
```

```python
# Change pen to notebook
shopping_list[3]="Notebook"
print(shopping_list)
```

```python
# Delete the least importnat item and prrint the final shopping list
shopping_list.remove("Football")
print("Final shopping list :",shopping_list)
```
