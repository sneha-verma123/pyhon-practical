```python
#QI. Write a Python program to check if a player Lionel Messi has more than IO achievements. If the condition is true, print the F*ayer's name, sport, and achievements else print does not have
#more than 10 achievements
players = [
    ["Serena Williams", "Tennis", 23],
    ["Lionel Messi", "Soccer", 7],
    ["Michael Phelps", "Swimming", 23],
    ["Usain Bolt", "Athletics", 8],
    ["Roger Federer", "Tennis", 20],
    ["Cristiano Ronaldo", "Soccer", 5]
]

for player in players:
    if player[0] == "Lionel Messi":
        if player[2] > 10:
            print(player[0], player[1], player[2])
        else:
            print("Lionel Messi does not have more than 10 achievements.")

```

```python
#Q4. Write a Python program in a Jupyter Notebook that takes an integer input (1 to 9999) from the user and determines the following using branching statements:
#1. If the number is even or odd
#2. If the number is a palindrome (reads the same forward and backward).
#3. If the number is divisible by the Sum Of its digits.
num = int(input("Enter an integer between 1 and 9999: "))

# Check even or odd
if num % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")

# Check palindrome
if str(num) == str(num)[::-1]:
    print("The number is a palindrome.")
else:
    print("The number is not a palindrome.")

# Calculate sum of digits
digit_sum = sum(int(digit) for digit in str(num))

# Check divisibility by sum of digits
if num % digit_sum == 0:
    print("The number is divisible by the sum of its digits.")
else:
    print("The number is not divisible by the sum of its digits.")
```

```python
#07. Write a while loop to copy the strings 'orange'of the list squares to the list new_squares .Stop and exit the loop if the value on the list is not
#'orange' 
squares = ['red', 'yellow', 'green', 'purple', 'blue']

new_squares = []

i = 0

while i < len(squares):
    if squares[i] != 'orange':
        break
    new_squares.append(squares[i])
    i += 1

print(new_squares)
```

```python
#08. The following is a list of animals in a National Zoo
# Animals =["lion","giraffe","gorilla","parrots","crocodile","deer"] Your brother needs to write an essay on the animals whose names are made of 7 letters. Help him
#find those animals through a while loop and create a separate list of such animals.
Animals = ["lion", "giraffe", "gorilla", "parrots", "crocodile", "deer", "swan"]

seven_letter_animals = []

i = 0

while i < len(Animals):
    if len(Animals[i]) == 7:
        seven_letter_animals.append(Animals[i])
    i += 1

print(seven_letter_animals)


```
