```python
# CREATE AN EMPTY DICTIONARY
inventory={}
```

```python
# Store the first product details in variables 
product_name="mobile phone"
product_quantity= 5
product_price= 20000
product_release_year= 2020
print("Product name:",product_name)
print("Product quantity:",product_quantity)
print("Product price:",product_price)
print("Pproduct release year:",product_release_year)
```

```python
# Add details in inventory
inventory["Product No.1"]={
    "ProductName":product_name,
    "ProductQuantity":product_quantity,
"ProductPrice":product_price,
    "ProductReleaseYear": product_release_year}
print(inventory)
```

```python
# Store the second product details in variables
product_name="laptop"
product_quantity= 2
product_price= 150000
product_release_year= 2024
print("Product name:",product_name)
print("Product quantity:",product_quantity)
print("Product price:",product_price)
print("Pproduct release year:",product_release_year)
    
```

```python
# Add second product details in inventory
inventory["Product No.2"]={
    "ProductName":product_name,
    "ProductQuantity":product_quantity,
"ProductPrice":product_price,
    "ProductReleaseYear": product_release_year}
print(inventory)
```

```python
# Display the products present in inventory 
print("Products in inventory:")
print(inventory)
```

```python
# Check if release years are present in inventory
print("Product No.1 release year:",
      inventory["Product No.1"]
      ["ProductReleaseYear"])
print("Product No.2 release year:",
      inventory["Product No.2"]
      ["ProductReleaseYear"])
```

```python
# Delete release years of both products
del inventory["Product No.1"]
["ProductReleaseYear"]
del inventory["Product No.2"]
["ProductReleaseYear"]
```

```python
# Create a tuple of product prices
prices=(250,300,150,400,100,350,200)
print(prices)
```

```python
# Find highest and lowest price
print("Highest price:",max(prices))
print("Lowest price:",min(prices))
```

```python
# Calculate total sum 
print("Total Sum:",sum(prices))
```

```python
# Convert tuple into sorted list 
sorted_prices=sorted(prices)
print("SortedList:",sorted_prices)
```

```python
# Try to modify an element in  the tuple 
try:
    prices[0]=500
except TypeError:
    print("Tuple cannot be modified because tuples are immutable")
```
